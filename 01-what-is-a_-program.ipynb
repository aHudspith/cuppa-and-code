## 2. A program is a list of instructions, run in order

**Predict:** in what order will these three lines appear?

```python
print("Kettle on")
print("Teabag in mug")
print("Pour the water")
```

**Modify it:** in the cell below, swap the order of the lines (cut and paste one line above another), then run it.

Python follows your new order exactly. It does not know that pouring water before adding a teabag is odd. It only knows what you told it.

```python
print("Kettle on")
print("Teabag in mug")
print("Pour the water")
```

## 3. Computers do not guess

**Predict:** how many times will "Add milk" appear?

```python
print("Add milk")
print("Add milk")
print("Add milk")
```

Three times, because that is what the program says. Computers do exactly what you tell them, even when it is a mistake. (Later in the series you will meet *loops*, which repeat things without copy and paste.)

## 4. The notebook habit: Restart and Run All

These notebooks **remember** things between cells. That is useful, but because you can run cells in any order, it can also trick you.

The next two cells use something called a *variable*. We will cover those properly in post 3. For now, just trust that the first cell makes the notebook remember the word "tea", and the second cell shows it.

```python
drink = "tea"
```

```python
print(drink)
```

**Experiment:**
1. Run the two cells above, in order. You should see `tea`.
2. In the menu, choose **Runtime > Restart session**. This wipes the notebook's memory.
3. Now run **only** the second cell, `print(drink)`.

You will see red error text. That is your first error, and it is nothing to panic about. The notebook forgot `drink` because you skipped the cell that creates it. We will learn to read errors properly in post 2.

4. Run both cells in order again, and it works.

**The habit:** whenever your notebook acts strangely, choose **Runtime > Restart session and run all**. It clears the memory and runs everything from the top, in order.
